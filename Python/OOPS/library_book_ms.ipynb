Goal: Create a class to manage book records in a library.

Requirements:

Class LibraryBook

Class attributes: library_name, location

Instance attributes: title, author, isbn, __availability (private)

Methods:

display() – shows all book info

change_availability(status) – sets availability (True/False)

is_available() – returns the current availability

@staticmethod def format_isbn(isbn) – formats the ISBN number nicely**
