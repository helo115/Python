**Objective:

Create a class Laptop to represent laptops in a tech store inventory system.

📦 Class: Laptop
Class Attributes:

store_name – name of the electronics store (e.g., "Tech World")

location – city name (e.g., "New York")

Instance Attributes (set via constructor):

brand – brand of the laptop (e.g., "Dell")

model – model name or number (e.g., "XPS 13")

price – price of the laptop (float)

__serial_number – private attribute, e.g., "SN123456789"

🧰 Methods:
✅ Instance Method:

display()

Prints all details of the laptop

Should include discounted price

Should also show the category of the laptop based on price

✅ Getter Method:

get_serial_number()

Returns the private serial number

✅ Setter Method:

update_serial_number(new_serial)

Updates the private serial number

✅ Static Method:

apply_discount(price)

Apply 15% discount to the price and return discounted price

✅ Static Method:

classify_laptop(price)

Classify laptop based on price:

Below $500 → "Entry Level"

$500–$1000 → "Mid-range"

Above $1000 → "High-End"

```python
from os import stat
class Laptop():
  # class Attribute
  store_name="Tech World"
  location="California"
  inventory=[]
  # constructor
  def __init__(self,brand,model,price,serial_no):
    self.brand=brand
    self.model=model
    self.price=price
    self.__serial_no=serial_no


    Laptop.inventory.append([self])
 # instance method
  def display(self):
    print("-------Laptop Details-------")
    print(f"The brand name of Laptop is {self.brand}")
    print(f"The model  of Laptop is {self.model}")
    print(f"The price of Laptop is {self.price}")
    print(f"The   Serial no of Laptop is {self.__serial_no}")
    print(f"After Discount the price is {self.apply_discount(self.price)}")
    print(f"The Category of laptop is {self.classify_laptop(self.price)}")
    print("The Record history of laptop is ")
    self.laptop_record()
    print("-----------------------------")
# Setter method ------------> instance method
  def update_serial_no(self,new_serial_no):
    self.__serial_no=new_serial_no
# Getter method-------------> instance method
  def get_serial_number(self):
    return self.__serial_no
# static method
  @staticmethod
  def apply_discount(price):
    discount=15/100*price
    discounted_price=price-discount
    return discounted_price
# static method
  @staticmethod
  def classify_laptop(price):
    if price>=25000 and price<60000:
      return 'Entry Level '
    elif price>=60000 and price<100000:
      return 'Mid-Range'
    elif price>=100000:
      return 'High-End'
    else:
      return 'Secound hand Entry level laptop'
  def laptop_record(self):
    for i in Laptop.inventory:
      print(i)
```
