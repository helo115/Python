**Do private the data OR Encapsulation**

```python
from os import access
class Studentrecord():
  # Class attribute
    institution="SMIT"
    campus="Hajiyani"

    def __init__(self,name,rollno,course,marks):
      self.name=name # Public
      self.rollno=rollno # Public
      self.course=course # Public
      self.__marks=marks # Private
    def display(self):
      print(f"institute:{self.institution}")
      print(f"Campus:{self.campus}")
      print("Name:",self.name)
      print("Rollno:",self.rollno)
      print("Course:",self.course)
      print("Marks:",self.__marks)
      print(f"Percentage: {Studentrecord.perc(Studentrecord.acess(self))}")
      print(f"Grade:{self.display_grade()}")

    # Setter method (To modify private data)
    def change(self,new_marks):
      self.__marks=new_marks
    # Getter method(To accesss privae data)
    def acess(self):
      return self.__marks
    # helper method or static method which only convert
    @staticmethod
    def perc(marks):
      percentage=marks*100/100
      return percentage
    @ staticmethod
    def grade(marks):
      if marks>=80 and marks<100:
        print("A+ grade")
      if marks>=70 and marks<80:
        print("A Grade")
      if marks>=60 and marks<70:
        print("B Grade")
      if marks>=50 and marks<60:
        print("C Grade")
      if marks>=40 and marks<50:
        print("C Grade")
      if marks<40:
        print("Fail")

    def display_grade(self):
      grade=self.grade(self.__marks)
      return grade
```
