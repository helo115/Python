**🏷️ Class Name:

BankAccount

🔧 Class Attributes (Shared by all objects):

bank_name → "National Trust Bank" (default)

interest_rate → 4.5 (default, as percentage for savings account)

📌 Instance Attributes (Unique per object):

account_holder → Name of the person

account_no → Unique account number

account_type → "savings" or "current"

__balance → Private, starting balance (e.g., 10000)

🧠 Methods (What your class should do):
1. display()

Purpose: Display all the account details.

Should include: name, account number, type, and current balance.

2. deposit(amount)

Purpose: Add money to the account.

Condition: Amount must be positive.

Effect: Increase balance.

3. withdraw(amount)

Purpose: Withdraw money from the account.

Conditions:

Amount must be positive.

Must not exceed available balance.

Effect: Decrease balance.

4. get_balance()

Purpose: Return the private balance value.

5. change_account_type(new_type)

Purpose: Change account type.

Only allow "savings" or "current" as valid types.

6. @staticmethod calculate_interest(balance, rate)

Purpose: Calculate yearly interest amount.

Formula: (rate / 100) * balance

Note: Default rate is class attribute interest_rate, but allow any rate to be passed.

```python
class BankAccount():
  # class Attribute
  bank_name = "National Trust Bank"
  interest_rate = 4.5

  #constructor
  def __init__(self,account_holder,account_no,account_type,balance):
    self.account_holder=account_holder
    self.account_no=account_no
    if account_type=="Saving" or account_type== "Current":
      self.account_type=account_type
    elif account_type=="saving"or account_type=="current":
      account_type.capitalize()
      self.account_type=account_type
    else:
       print("This type of Account is not present please choose from (Saving or Current )")
    self.__balance=balance
    self.transaction_historyy=[]
  # instance method
  def display(self):
    print("-----------Account Datails-----------")
    print(f"The Name of Account holder is {self.account_holder}")
    print(f"The Account no of Account holder is {self.account_no}")
    print(f"The Account Type is  {self.account_type}")

    print(f"The Current Balance in the Account is {self.__balance}")
    print(f"The Yearly Interest will been decuted is {self.calculate_interest(self.get_balance())}")
    print(f"Transaction history is ")
    self.transaction_history()
    print("--------------------------------------")
  # Setter method ----------> instance method
  def deposit(self,amount):
    if amount>=1:
      self.__balance=self.__balance+amount
      self.transaction_historyy.append(f"Deposited {amount}")
    else:
      print("The Deposit must be greater 0")
# Setter method ----------> instance method
  def withdraw(self):
    amount=int(input("Enter the amount you want to withdraw"))
    if amount>=1:
      if self.__balance>=amount:
        s=amount
        self.__balance=self.__balance-amount
        self.transaction_historyy.append(f"With Drawn {amount}")
        return s
      else:
        print(f"You have only {self.__balance} in your account")
        print("Insufficient Amount in the account to with draw")

    else:
      print("Withdraw Amount always be greather than 0")

# gatter method -------------> instance method
  def get_balance(self):
    return self.__balance
# setter method -------------> instance method
  def change_account_type(self,new_type):
    if new_type=="saving"or new_type=="current":
      new_type.capitalize()
      if self.account_type==new_type:
        print("The Account is already of this type ")
      else:
        self.account_type=new_type
        print("Successfully Updated")
    elif new_type=="Saving" or new_type=="Current":
      if self.account_type==new_type:
        print("The Account is already of this type ")
      else:
        self.account_type=new_type
        print("Successfully Updated")

    else:
      print("This type of Account is not present please choose from (Saving or Current )")
  # static method
  @staticmethod
  def calculate_interest(balance):
    rate=float(input("Enter the Interest Rate:"))
    interest=(rate/100)*balance

    return interest

  # class method
  @classmethod
  def change_bank_name(cls,new_name):
    cls.bank_name=new_name

  def transaction_history(self):
    b=0
    for i in self.transaction_historyy:
      print(i)
```
