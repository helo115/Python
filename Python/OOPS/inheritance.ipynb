**Inheritance**

```python
class Parent():
  #Class Attribute
  Home="unknown"
  Campus="None"
  # constructor
  def __init__(self,eyecolor,haircolor,language):
    self.eyecolor=eyecolor
    self.haircolor=haircolor
    self.language=language
  def speak(self):
    return f"Parent can speak {self.language}"
```

```python
h=Parent("brown","Black","urdu")
```

```python
h.Home
```

```python
h.haircolor
```

```python
h.eyecolor
```

```python
h.language
```

```python
h.speak()
```

```python
class Child(Parent):
  def __init__(self,eyecolor,h_color,lang,courses):
    super().__init__(eyecolor,h_color,lang)
    self.coursee=courses
  def speak(self):
    return f"Child can speak {self.language}"
```
