**Polymerism** is a method in which if parent and child class have same function and its argument than child class method run

```python
class Parent():
  #class Attribute
  institute="TSLT"
  campus="street no 2"
  def __init__(self,name,language,course):
    self.name=name
    self.language=language
    self.course=course
  def Speak(self):
    return f"Parent can speak {self.language}"
```

```python
class Child(Parent):
  def speak(self):
    return f"child can speak {self.language}"
```

```python
p=Parent("Name","English","Pnnn")
```

```python
c=Child("None","English","Python")
```

```python
c.speak()
```

```python
p.Speak()
```
