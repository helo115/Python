```python
class Vechile():
  #class Attribute
  company="Toyota"
  Standard="High"
  def __init__(self,tire,engine_type,color,model,start):
    self.tire=tire
    self.engine_type=engine_type
    self.color=color
    self.model=model
    self.start=start
  from abc import ABC,abstractmethod
  @abstractmethod
  def Start(self):
    pass
  def display(self):
    print(f"The  Brand name is {self.company}")
    print(f"The Standard of vechicle  is {self.Standard}")
    print(f"The Tire Brand is {self.tire}")
    print(f"The Engine  is {self.engine_type}")
    print(f"The color is {self.color}")
    print(f"The model is {self.model}")
    print(f"The vechile is start by  {self.Start()}")
```

```python
class Bike(Vechile):
  def Start(self):
    return f"The Bike is Start by {self.start}"
```

```python
class Scooty(Vechile):
  def Start(self):
    return f"The Scooty is start by {self.start}"
```

```python
b=Bike("Service","Petrol","Black","2022","Kick")
```

```python
s=Scooty("Panther","Eletric","Pink","2025","Self")
```
