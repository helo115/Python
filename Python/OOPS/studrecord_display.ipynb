```python
class StudentRecords():
# class Attribute
  institution ="smit"
  campus="hajyani"

  def __init__(self,name,roll_no,course,marks):
  # instance attributes
    self.name = name
    self.roll_no = roll_no
    self.course = course
    self.__marks = marks

  # instance method
  def display(self):
    print("------institute detail-------")
    print(f"intitute: {self.institution}")
    print(f"campus: {self.campus}")
    print("-------student detail-------")
    print(f"name: {self.name}")
    print(f"roll_no: {self.roll_no}")
    print(f"course: {self.course}")
    print(f"marks: {self.__marks}")
```

```python
s1=StudentRecords("taha",1234,"ai",90 )
```

```python
s2=StudentRecords("mahad",2345,"ai",85)
```

```python
s1.display()
```
